# Dataset Generation and Manipulation

This notebook is used for experimenting with dataset generation and manipulation techniques for employee attrition data. It allows for interactive exploration of data generation methods and the addition of new variables.

```python
import pandas as pd
import numpy as np
import random

def generate_synthetic_data(num_records=2000, random_seed=42):
    """Generate synthetic employee attrition data with additional variables."""
    np.random.seed(random_seed)
    random.seed(random_seed)

    departments = ["Sales", "Engineering", "Customer Success", "Operations", "Marketing", "Finance", "HR", "Legal"]
    grades = ["A", "B", "C", "D"]
    locations = ["HQ", "Cape Town", "Johannesburg", "Remote"]
    gender = ["Male", "Female"]

    data = {
        "employee_id": np.arange(1, num_records + 1),
        "age": np.random.randint(21, 63, num_records),
        "tenure_months": np.random.randint(3, 180, num_records),
        "salary": np.random.normal(450000, 120000, num_records).clip(120000, 1200000),
        "department": np.random.choice(departments, num_records),
        "grade": np.random.choice(grades, num_records, p=[0.35, 0.4, 0.2, 0.05]),
        "location": np.random.choice(locations, num_records, p=[0.3, 0.35, 0.2, 0.15]),
        "gender": np.random.choice(gender, num_records),
        "performance_rating": np.random.choice([2, 3, 4, 5], num_records, p=[0.1, 0.35, 0.4, 0.15]),
        "engagement_score": np.random.normal(65, 15, num_records).clip(10, 100),
        "overtime_hours_month": np.random.randint(0, 45, num_records),
        "absences_12m": np.random.randint(0, 20, num_records),
        "training_hours_last_yr": np.random.randint(0, 80, num_records),
        "num_promotions": np.random.randint(0, 5, num_records),
        "pay_change_12m_pct": np.random.normal(3, 6, num_records).clip(-15, 25),
        "market_delta_pct": np.random.normal(0, 8, num_records).clip(-20, 30),
        "remote_ratio": np.random.randint(0, 100, num_records),
        "leave_balance_days": np.random.randint(0, 35, num_records),
        "attrition_12m": np.random.randint(0, 2, num_records),
        "attrition_3m": np.random.randint(0, 2, num_records)
    }

    df = pd.DataFrame(data)
    return df

# Generate synthetic dataset
synthetic_data = generate_synthetic_data(num_records=5000)
synthetic_data.head()
```

## Saving the Generated Dataset

After generating the dataset, we can save it to a CSV file for further use in the project.

```python
synthetic_data.to_csv('../data/sample_attrition_data.csv', index=False)
len(synthetic_data)
```
